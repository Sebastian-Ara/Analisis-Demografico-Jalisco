## Índice de Whipple
El Índice de Whiple mide la preferencia por declarar la edad en multiplos de 5, nos ayuda a evaluar que tan bien están los datos respecto a la edad y se calcula de la siguiente manera:

$I.W=\frac{5\times(P_{25}+P_{30}+...+P_{60})}{\displaystyle \sum_{23}^{62} P_x }\times 100$

Donde $P_x$ es la población de x años.
Se tiene la siguiente escala:

[100,105) : Datos Muy Precisos.

[105,110): Datos Precisos.

[110,125) : Datos Aproximados.

[125,175): Datos Deficientes

[175,50]: Datos Deficientes

```python
#Suma de todas las edades de los 23 hasta los 62 2010
sQH_2010 = sum(individual2010.iloc[2:40:5 , 2 ].values)
sQM_2010 = sum(individual2010.iloc[2:40:5 , 3 ].values)

#Suma de todas las edades de los 23 hasta los 62 2020
sQH_2020 = sum(individual2020.iloc[2:40:5 , 2 ].values)
sQM_2020 = sum(individual2020.iloc[2:40:5 , 3 ].values)

#Suma de las edades multiplos de 5 del 2010
sHI_2010 = sum(individual2010.iloc[0:40 , 2].values)
sMI_2010 = sum(individual2010.iloc[0:40 , 3].values)

#Suma de las edades multiplos de 5 del 2020
sHI_2020 = sum(individual2020.iloc[0:40 , 2].values)
sMI_2020 = sum(individual2020.iloc[0:40 , 3].values)
```

```python
#Calcular el indice
IW_H_2010 = (((5)*(sQH_2010))/sHI_2010)*100
IW_M_2010 = (((5)*(sQM_2010))/sMI_2010)*100
IW_H_2020 = (((5)*(sQH_2020))/sHI_2020)*100
IW_M_2020 = (((5)*(sQM_2020))/sMI_2020)*100
```

# Índice de Myers 
El indice de Myers es un indicador que mide la preferencia que tienen las personas por declarar sus edades terminadas en el digito $j$, con $j \in \mathbf{0}, \mathbf{1}, 2, \ldots, 8,9$
Se calcula de la siguiente manera:
$$
I M=\sum_{j=0}^9\left|M_j\right|
$$
donde,

$\begin{gathered}M_j=\left(\frac{a_j P_j+a_j^{\prime} P_j^{\prime}}{\sum_{k=0}^9\left(a_k P_k+a_k^{\prime} P_k^{\prime}\right)}-\frac{1}{10}\right) \cdot 100 \\ a j=j+1 \quad a j^{\prime}=9-j,\end{gathered}$

para cada $j$ en $0,1, \ldots, 9$

```python
def IndiceMyers(df, state, year):
# Obtención de ajPj + aj'Pj' y Mj para hombres
    
    #ajPj + aj'Pj'
    suma_ajPj_H = []
    for j in range(10):
        aj = j + 1
        aj_prima = 9 - j

        Pj = sum(df.iloc[10+j : 61+j : 10 , 1 ].values)
        Pj_prima = sum(df.iloc[20+j : 71 + j : 10 , 1].values)
        suma_ajPj = (aj * Pj) + (aj_prima * Pj_prima)
        suma_ajPj_H.append(suma_ajPj)
     
    # Mj
    Mj_H = []
    for i in suma_ajPj_H:
        Mj = ((i/sum(suma_ajPj_H))-(1/10))*100
        Mj_H.append(Mj)
    
    # Obtención de ajPj + aj'Pj' y Mj para mujeres  

    # ajPj + aj'Pj'
    suma_ajPj_M = []

    for j in range(10):
        aj = j + 1
        aj_prima = 9 - j

        Pj = sum(df.iloc[10+j : 61+j : 10 , 2].iloc)
        Pj_prima = sum(df.iloc[20+j : 71 + j : 10 , 2].iloc)
        suma_ajPj = (aj * Pj) + (aj_prima * Pj_prima)
        suma_ajPj_M.append(suma_ajPj)
    
    # Mj
    Mj_M = []
    for i in suma_ajPj_M:
        Mj = ((i/sum(suma_ajPj_M))-(1/10))*100
        Mj_M.append(Mj)
    
    # Cálculo del Indice de Myers
    
    IM_H = 0
    IM_M = 0

    for Mj in Mj_H:
        IM_H += abs(Mj)
    for Mj in Mj_M:
        IM_M += abs(Mj)
    
    print(f'El índice de Myers para la población masculina de {state} en el censo del {year} fue de {IM_H}')
    print(f'El índice de Myers para la población femenina de {state} en el censo del {year} fue de {IM_M}')
```

# Índice de las Naciones Unidas 
El índice mide las irregularidades en la declaración de la edad y el sexo por lo que no podemos hacer análisis separados por sexo. Se calcula de la siguiente manera:

$$
INU = \frac{1}{15}\left[\sum_{k = 5}^{75}\ _5CEH_k + \sum_{k = 5}^{75} \ _5CEM_k + 3\sum_{k = 5}^{75} \Delta_5IM_k\right]$$

donde

$$_5CEH_k = \left|\frac{2\ _5P_k^H}{_5P_{k-5}^H +\ _5P_{k+5}^H} - 1\right| \cdot 100 \qquad _5CEM_k = \left|\frac{2\ _5P_k^M}{_5P_{k-5}^M +\ _5P_{k+5}^M} - 1\right| \cdot 100$$

$$\Delta_5IM_k = |\ _5IM_k -\ _5IM_{k-5}|$$

$$_5IM_k =\frac{\ _5P_k^H}{\ _5P_k^M} \cdot 100 $$

con 

$ _5P^M_k=$ Población de mujeres en el intervalo de edades [x,x+5) 

$ _5P^H_k=$ Población de hombres en el intervalo de edades [x,x+5) 

Con la siguiente escala de valores:


- [0,20) Información Satisfactoria.


- [20,40) Información de Calidad Intermedia.


- [40, $\infty$ ) Información Deficiente.





```python
def IndiceNacionesUnidas(df, state, year):
    
    # Rediseño de los dataFrames
    df = df.rename(columns = {'Unnamed: 0':'Edad'})
    df = df.drop(['Total'], axis = 1)

    CEH = []
    CEM = []
    DeltaIM =  []

    for i in range (1,16): 
        # 1 porque tengo que contemplar el quinquenio pasado y 
        # 16 porque el último quinquenio de 75 a 79 años es el índice 15
        pobH_presente = df.iloc[:, 1].values[i] # el primer parámetro del iloc, abarca toda la columna de hombres
        pobH_pasada = df.iloc[:, 1].values[i-1]
        pobH_futura = df.iloc[:, 1].values[i+1]

        pobM_presente = df.iloc[:, 2].values[i]
        pobM_pasada = df.iloc[:, 2].values[i-1]
        pobM_futura = df.iloc[:, 2].values[i+1]
        
        # Para la construcción del delta IM

        IM_presente = ((pobH_presente)/(pobM_presente))*100
        IM_pasado = ((pobH_pasada)/(pobM_pasada))*100

        CEH.append(abs(((2*pobH_presente)/(pobH_pasada + pobH_futura))-1)*(100))
        CEM.append(abs(((2*pobM_presente)/(pobM_pasada + pobM_futura))-1)*(100))
        DeltaIM.append(abs(IM_presente - IM_pasado))
        
        INU = (1/15)*(sum(CEH)+sum(CEM)+ (3*sum(DeltaIM)))
        
    print(f'El Indice de Naciones Unidas de la población de {state} de acuerdo con el censo del {year} fue de {INU}')
```

```python
IndiceNacionesUnidas(quinquenal2010, 'Jalisco', 2010)
```

```python
IndiceNacionesUnidas(quinquenal2020, 'Jalisco', 2020)
```

Con esto vemos que para ambos censos la información obtenida del estado de Jalisco es satisfactoria. Es preciso destacar que se muestra una mejora importante desde 2010 hasta 2020. Esta mejora es una prueba directa del refinamiento en los métodos utilizados por el INE para recopilar datos demográficos y sociales; puede atribuirse a una mayor precisión en la recopilación de datos o a posibles avances tecnológicos que faciliten este proceso, incluso un diferente presupuesto o distribución de este. 

  

Sin embargo, al comparar los resultados del Índice de las Naciones Unidas con el Índice de Whiple, se observa una diferencia en la calidad de la evaluación de los datos. Mientras que el Índice de las Naciones Unidas califica la información como "satisfactoria", el Índice de Whiple la clasifica como "datos aproximados", lo que sugiere una evaluación menos precisa.  

Una posible explicación para esta diferencia podría ser los factores que tiene en cuenta cada Índice pues el Índice de las Naciones Unidas además de considerar la edad, toma en cuenta el género, valor del que es más difícil obtener datos erróneos. Así que la certeza al obtener el género compensaría las deficiencias de obtención de la edad y se reflejaría en una mayor precisión de obtención de datos. 

# Piramides de población individuales
Es un doble histograma de la distribución de población de hombres y mujeres por edades. Se pueden identificar 4 tipos: progresiva, regresiva, estacionaria y desequilibrada. 


## Piramide de población 2010

```python
def PiramidePoblacional(df, state, year):
    
    df = df.rename(columns={'Unnamed: 0': 'Edad'}).drop('Total', axis=1)
    fig, ax = plt.subplots(figsize = (24, 20)) # figsize(ancho, altura)
    ax.barh(df['Edad'], -df.iloc[:, 1], color = 'blue', label = 'Hombres')
    ax.barh(df['Edad'], df.iloc[:, 2], color = 'fuchsia', label = 'Mujeres')

    ax.set_xlabel('Hombres | Mujeres')
    ax.set_title(f'Pirámide de Población de {state} en {year}', fontsize = 20)
    ax.legend(fontsize = 16)
    
    #Importar imagenes
    img_hombre = plt.imread('hombre.png')  # Cambia 'hombre.png' por la ruta de tu imagen de hombre
    img_mujer = plt.imread('mujer.png')  # Cambia 'mujer.png' por la ruta de tu imagen de mujer

    # Definir la ubicación de las imágenes
    xy_hombre = (0.2, 0.8)  # Coordenadas para la imagen del hombre
    xy_mujer = (0.8, 0.8)  # Coordenadas para la imagen de la mujer

    # Agregar imágenes al gráfico
    ax_hombre = AnnotationBbox(OffsetImage(img_hombre, zoom=0.5), xy=xy_hombre, xycoords='axes fraction', frameon=False)
    ax_mujer = AnnotationBbox(OffsetImage(img_mujer, zoom=0.5), xy=xy_mujer, xycoords='axes fraction', frameon=False)
    ax.add_artist(ax_hombre)
    ax.add_artist(ax_mujer)
    
    
    plt.show()
```
